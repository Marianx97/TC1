## Resolución

$\displaystyle s(t) = \begin{cases}
  50mA.u(t); \quad 0<t<T_1;\\
  0; \quad T_1<t<T \\
\end{cases}$

$\displaystyle s_{me}(t) = \frac{1}{T} . \int_{0}^{T} s(t).dt = \frac{1}{T} . \int_{0}^{T_1} 50mA.u(t).dt = \frac{50mA}{T} . [t]^{T_1}_{0} = \frac{50mA}{T} . [T_1-0] = 50mA . \frac{T_1}{T} $

Si $s_{me}(t) = 10mA$ entonces el ciclo de actividad será:

$\displaystyle 10mA = 50mA . \varepsilon$

$\displaystyle \varepsilon = \frac{10mA}{50mA} = \frac{1}{5} = 0,2$

El ciclo de actividad será el $20\%$ del período.
