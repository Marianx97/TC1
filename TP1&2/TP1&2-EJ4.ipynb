## Resolución

### a)

$\displaystyle s(t) = \frac{S_m}{T} . P(t);\quad n.T<t<(n+1).T;\quad n=\{0,1,2,...\}$

$\displaystyle s_{me}(t) = \frac{1}{T} . \int_{0}^{T} s(t).dt = \frac{1}{T} . \int_{0}^{T} \frac{S_m}{T} . P(t).dt = \frac{S_m}{T^2}.\int_{0}^{T} t.dt$

$\displaystyle s_{me}(t) = \frac{S_m}{T^2}.\left[\frac{t^2}{2}\right]_{0}^{T} = \frac{S_m}{2.T^2}.[T^2-0^2] = \frac{S_m}{2}$

$\displaystyle s_{ef}(t) = \sqrt{P_s(t)} = \sqrt{\frac{1}{T} . \int_{0}^{T} |s(t)|^2.dt} = \sqrt{\frac{1}{T} . \int_{0}^{T}\frac{S_m^2}{T^2}.t^2.dt} = \sqrt{\frac{S_m^2}{T^3} . \int_{0}^{T}t^2.dt}$

$\displaystyle s_{ef}(t) = \sqrt{\frac{S_m^2}{T^3} . \left[\frac{t^3}{3}\right]_{0}^{T}} = \sqrt{\frac{S_m^2}{3.T^3} . [T^3-0^3]} = \frac{S_m}{\sqrt{3}}$

$\displaystyle f_{|me|} = \frac{S_m}{\frac{S_m}{2}} = 2;\quad f_c = \frac{S_m}{\frac{S_m}{\sqrt{3}}} = \sqrt{3};\quad f_f = \frac{2}{\sqrt{3}}$

### b)

$\displaystyle s(t) = \begin{cases}
  s.u(t); \quad 0<t<T_1 \\
  -s.u(t); \quad T_1<t<T \\
\end{cases}$

$\displaystyle s_{me}(t) = \frac{1}{T} . \int_{0}^{T} s(t).dt = \frac{1}{T} . \left(\int_{0}^{T_1} s.u(t).dt+\int_{T_1}^{T} -s.u(t).dt\right) = \frac{1}{T} . \left([s.t]^{T_1}_0-[s.t]^{T}_{T_1}\right)$

$\displaystyle s_{me}(t) = \frac{s}{T} . [(T_1-0)-(T-T_1)] = \frac{s.(2.T_1-T)}{T}$

$\displaystyle s_{ef}(t) = \sqrt{\frac{1}{T} . \int_{0}^{T} |s(t)|^2.dt} = \sqrt{\frac{1}{T} . \int_{0}^{T} s^2.(u(t))^2.dt} = \sqrt{\frac{s^2}{T} . \int_{0}^{T} 1^2.dt} $

$\displaystyle s_{ef}(t) = \sqrt{\frac{s^2}{T} . [t]_{0}^{T}} = \sqrt{\frac{s^2}{T} . (T-0)} = \sqrt{s^2} = s$

$\displaystyle s_{|me|} = \frac{1}{T} . \int_{0}^{T} |s(t)|.dt = \frac{1}{T} . \int_{0}^{T} s.u(t).dt = \frac{s}{T} . \int_{0}^{T} 1.dt = \frac{s}{T} . [t]_{0}^{T} = \frac{s}{T} . (T-0) = \frac{s}{T} . T = s$

$\displaystyle f_{|me|} = \frac{s}{s} = 1;\quad f_c = \frac{s}{s} = 1;\quad f_f = \frac{1}{1} = 1$

### c)

$\displaystyle s(t) = \begin{cases}
  s_m.\sin{\left(\omega_0.t\right)}; \quad 0<t<\frac{T}{2}; \quad \frac{2\pi}{T} \\
  0; \quad \frac{T}{2}<t<T \\
\end{cases}$

$\displaystyle s_{me}(t) = \frac{1}{T} . \int_{0}^{T} s(t).dt = \frac{1}{T} . \int_{0}^{\frac{T}{2}} s_m.\sin{\left(\omega_0.t\right)}.dt = \frac{s_m}{T} . \int_{0}^{\frac{T}{2}} \sin{\left(\omega_0.t\right)}.dt = \frac{s_m}{T.\omega_0} . \int_{0}^{\frac{T}{2}} \sin{(u)}.du$

$\displaystyle s_{me}(t) = \frac{s_m}{T.\frac{2\pi}{T}} . \left(-\cos{(\omega_0.t)}\right)^{\frac{T}{2}}_{0}$

$\displaystyle s_{me}(t) = \frac{s_m}{2\pi} . \left[-\cos{\left(\omega_0.\frac{T}{2}\right)}-(-\cos{(\omega_0.0)})\right] = \frac{s_m}{2\pi} . \left[-\cos{(\pi)}-(-\cos{(0)})\right] = \frac{s_m}{2\pi} . [-(-1)-(-1)] = \frac{s_m}{2\pi}.2$

$\displaystyle s_{me}(t) = \frac{s_m}{\pi}$

$\displaystyle s_{ef}(t) = \sqrt{\frac{1}{T} . \int_{0}^{T} |s(t)|^2.dt} = \sqrt{\frac{1}{T} . \int_{0}^{\frac{T}{2}} s_m^2.\sin^2{(\omega_0.t)}.dt} = \sqrt{\frac{s_m^2}{T} . \int_{0}^{\frac{T}{2}} \sin^2{(\omega_0.t)}.dt}$

$\displaystyle s_{ef}(t) = \sqrt{\frac{s_m^2}{T} . \left[\frac{t}{2} - \frac{\sin{(\omega_0.t)}}{4\omega_0} \right]_{0}^{\frac{T}{2}}} = \sqrt{\frac{s_m^2}{T} . \left( \left[\frac{\frac{T}{2}}{2} - \frac{\sin{(\omega_0.\frac{T}{2})}}{4\omega_0}\right] - \left[\frac{0}{2} - \frac{\sin{(\omega_0.0)}}{4\omega_0}\right]\right)}$


$\displaystyle s_{ef}(t) = \sqrt{\frac{s_m^2}{T} . \frac{T}{4}} = \frac{s_m}{2}$

$\displaystyle f_{|me|} = \frac{s_m}{\frac{s_m}{\pi}} = \pi;\quad f_c = \frac{s_m}{\frac{s_m}{2}} = 2;\quad f_f = \frac{\pi}{2}$
