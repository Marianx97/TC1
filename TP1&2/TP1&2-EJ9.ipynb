## Resolución

### a)

$\displaystyle v_c(t) = v(t) - v_{R1}(t) = v(t) - i(t) . R1$

$\displaystyle i(t) = i_c(t) + i_{R2}(t)$

$\displaystyle i(t) = C. \frac{d(v_c(t))}{dt} + \frac{v_c(t)}{R2}$

$\displaystyle i(t) = C. \frac{d(v(t) - i(t) . R1)}{dt} + \frac{v(t) - i(t) . R1}{R2}$

$\displaystyle i(t) = C.v'(t) - C.R1.i'(t) + \frac{1}{R2}.v(t) - \frac{R1}{R2}.i(t)$

$\displaystyle i(t) + C.R1.i'(t) + \frac{R1}{R2}.i(t) = C.v'(t) + \frac{1}{R2}.v(t)$

$\displaystyle C.R1.i'(t) + \left(1 + \frac{R1}{R2}\right).i(t) = C.v'(t) + \frac{1}{R2}.v(t)$

### b)

$\displaystyle i_l(t) = i(t) - i_{R2}(t)$

$\displaystyle i_l(t) = i(t) - \frac{v(t)-v_{R1}(t)}{R2}$

$\displaystyle i_l(t) = i(t) - \frac{v(t)}{R2} + \frac{R1}{R2} . i(t)$

$\displaystyle i_l(t) = \left(1 + \frac{R1}{R2} \right) . i(t) - \frac{v(t)}{R2}$

$\displaystyle v(t) = v_{R1}(t) + v_l(t)$

$\displaystyle v(t) = i(t).R1 + L.\frac{d(i_l(t))}{dt}$

$\displaystyle v(t) = i(t).R1 + L.\left[\left(1 + \frac{R1}{R2} \right) . i'(t) - \frac{v'(t)}{R2}\right]$

$\displaystyle \frac{L}{R2}.v'(t) + v(t) = L.\left(1 + \frac{R1}{R2}\right) . i'(t) + R1.i(t)$
