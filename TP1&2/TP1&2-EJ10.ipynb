## Resolución

$\displaystyle \begin{cases}
  V_1 + V_{R1} + V_L = 0 \\
  V_2 + V_{R2} + V_L = 0 \\
  i_1 + i_2 = i_3
\end{cases} $

$\displaystyle \begin{cases}
  V_1 + i_1.R1 + L.\frac{d(i_3)}{dt} = 0 \\
  V_2 + i_2.R2 + L.\frac{d(i_3)}{dt} = 0 \\
  i_1 + i_2 = i_3
\end{cases} $
