# Resolución

$\displaystyle i_c (t) = C.\frac{dv(t)}{dt}$

Del gráfico:

$\displaystyle v(t) = \begin{cases}
  \rho(t) - 1; \quad 0<t<2 \\
  3.u(t); \quad 2<t<3\\
  2.u(t); \quad 3<t<4\\
  1.u(t); \quad 4<t<5\\
  -\rho(t) + 6; \quad 5<t<6 \\
\end{cases} $

Por lo tanto:

$\displaystyle i_c(t) = \begin{cases}
  1x10^{-6}.u(t); \quad 0<t<2 \\
  3x10^{-6}.\delta(t); \quad 2<t<3\\
  2x10^{-6}.\delta(t); \quad 3<t<4\\
  1x10^{-6}.\delta(t); \quad 4<t<5\\
  -1x10^{-6}.u(t); \quad 5<t<6 \\
\end{cases} $
