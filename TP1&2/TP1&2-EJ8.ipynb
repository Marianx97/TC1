## Resolución

### a)

$\displaystyle v(t) = v_{C1}(t) + v_{C2}(t)$

$\displaystyle 10.u(t) = \frac{1}{C1} . \int^{t}_{-\infty} i_c(t).dt + \frac{1}{C2} . \int^{t}_{-\infty} i_c(t).dt$

$\displaystyle 10.u(t) = \left(\frac{1}{C1}+\frac{1}{C2}\right) . \int^{t}_{-\infty} i_c(t).dt$

$d(\displaystyle 10.u(t)) = d\left(\left(\frac{1}{C1}+\frac{1}{C2}\right) . \int^{t}_{-\infty} i_c(t).dt\right)$

$\displaystyle 10.\delta(t).dt = \left(\frac{1}{C1}+\frac{1}{C2}\right).i_c(t).dt$

$\displaystyle i_c(t) = \left(\frac{10}{\frac{1}{C1}+\frac{1}{C2}}\right).\delta(t) = \left(\frac{10}{\frac{1}{2 \mu F}+\frac{1}{6\mu F}}\right).\delta(t)$

$\displaystyle i_c(t) = 15.\delta(t) [\mu A]$

### b)

$\displaystyle v_{C1}(t) = \frac{1}{C1} . \int^{t}_{-\infty} i_c(t).dt = \frac{1}{2\mu F} . \int^{t}_{-\infty} 15\mu A.\delta(t).dt = \frac{15 \mu A}{2 \mu F} . \int^{t}_{-\infty} \delta(t).dt = 7,5 V . u(t)$

$\displaystyle v_{C2}(t) = \frac{1}{C2} . \int^{t}_{-\infty} i_c(t).dt = \frac{1}{6\mu F} . \int^{t}_{-\infty} 15\mu A.\delta(t).dt = \frac{15 \mu A}{6 \mu F} . \int^{t}_{-\infty} \delta(t).dt = 2,5 V . u(t)$

### c)

Carga final en el capacitor:

$\displaystyle C1 = \frac{q_1(t)}{v_{C1}(t)} \to q_1(t) = C1 . v_{C1}(t) = 2\mu F . 7,5 V . u(t) = 15 \mu C .u(t)$

$\displaystyle C2 = \frac{q_2(t)}{v_{C2}(t)} \to q_2(t) = C2 . v_{C2}(t) = 6\mu F . 2,5 V . u(t) = 15 \mu C .u(t)$
