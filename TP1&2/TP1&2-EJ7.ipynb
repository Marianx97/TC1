## Resolución

### a)

$\displaystyle i(t) = I.u(t)$

$\displaystyle i_L(0^-) = I_0$

$\displaystyle i_L(t) = i(t) + i(0^-) = I.u(t) + I_0$

![grafico-a](img/tp1&2-ej7-graficos-a.JPG)


### b)

$\displaystyle i_L(t) = \frac{1}{L}.\int_{-\infty}^{t} v_L(t).dt \to d(i_L(t)) = d\left(\frac{1}{L}.\int_{-\infty}^{t} v_L(t).dt\right)$

$\displaystyle i_L(t) = I.u(t) + I_0$

$\displaystyle d(I.u(t) + I_0) = d\left(\frac{1}{L}.\int_{-\infty}^{t} v_L(t).dt\right)$

$\displaystyle I.\delta(t).dt + 0 = \frac{1}{L}.v_L(t).dt$

$\displaystyle I.\delta(t)= \frac{1}{L}.v_L(t)$

$\displaystyle v_L(t) = L.I.\delta(t)$

### c) y d)

Energía almacenada en un inductor:

$\displaystyle E = \frac{1}{2} . L . (i(t))^2$

Para $t < 0$:

$\displaystyle E = \frac{1}{2} . L . (I_0)^2$


Para $t > 0$:

$\displaystyle E = \frac{1}{2} . L . (I_0+I)^2$
